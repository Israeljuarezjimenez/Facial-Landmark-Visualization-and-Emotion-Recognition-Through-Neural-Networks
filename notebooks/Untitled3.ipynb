```python
import pandas as pd

# Función a minimizar
def h(x):
    return x**4 - 6*x**2 + 5*x

# Parámetros iniciales
a = 1
b = 3
delta = 0.01
phi = 0.618  # Relación áurea
alpha = 1 - phi

# Inicialización de las iteraciones
iterations = []
iteration = 1

while (b - a) > delta:
    # Calculamos x1 y x2
    x1 = a + alpha * (b - a)
    x2 = a + phi * (b - a)
    
    # Evaluamos la función en x1 y x2
    f_x1 = h(x1)
    f_x2 = h(x2)
    
    # Guardamos los resultados de esta iteración
    iterations.append([iteration, a, b, x1, x2, f_x1, f_x2])
    
    # Actualizamos el intervalo
    if f_x1 < f_x2:
        b = x2
    else:
        a = x1
    
    iteration += 1

# Crear un DataFrame para mostrar los resultados
columns = ["Iteración", "a", "b", "x1", "x2", "f(x1)", "f(x2)"]
df_iterations = pd.DataFrame(iterations, columns=columns)

# Mostrar los resultados al usuario
import ace_tools as tools; tools.display_dataframe_to_user(name="Iteraciones Método de la Sección Áurea", dataframe=df_iterations)
```
